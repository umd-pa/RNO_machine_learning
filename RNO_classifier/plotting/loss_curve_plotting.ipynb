```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator, AutoMinorLocator
data = pd.read_csv('/data/i3store/users/nmanic/classifier_consistency/ARZ2020/benchmark/runs/exp01/metrics.csv')
train_loss = data['train_loss']
val_loss = data['val_loss']
epochs = data['epoch']
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["cmr10", "Computer Modern Roman", "DejaVu Serif"],
    "mathtext.fontset": "cm",
    "axes.unicode_minus": False,
})

labelsize=18

fig, ax = plt.subplots(figsize=(6, 5))
ax.plot(epochs, train_loss, label="Training set"); ax.plot(epochs, val_loss, label="Validation set")
ax.set_xlabel("Epoch", fontsize=labelsize)
ax.set_ylabel("BCE Loss", fontsize=labelsize)
ax.legend(fontsize=12)
ax.tick_params(axis='both', which='major', labelsize=16)
#ax.grid(True)
# ax.xaxis.set_major_locator(MaxNLocator(nbins=6))
# ax.yaxis.set_major_locator(MaxNLocator(nbins=6))
# ax.xaxis.set_minor_locator(AutoMinorLocator(2))
# ax.yaxis.set_minor_locator(AutoMinorLocator(2))
# ax.grid(True, which='major', linestyle='-', linewidth=0.6, alpha=.7)
# ax.grid(True, which='minor', linestyle=':', linewidth=0.4, alpha=.7)
fig.tight_layout()
plt.show()


```
